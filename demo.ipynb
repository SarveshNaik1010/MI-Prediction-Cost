```python
import pandas as pd
import numpy as np

n = 5000

df = pd.DataFrame({
    'age': np.random.randint(18, 65, n),
    
    # Slightly imbalanced (realistic)
    'gender': np.random.choice(['Male', 'Female'], n, p=[0.55, 0.45]),
    
    'income': np.random.randint(30000, 120000, n),
    
    'children': np.random.choice([0,1,2,3,4], n, p=[0.25, 0.30, 0.20, 0.15, 0.10]),
    
    # More moderate, fewer extremes
    'exercise_level': np.random.choice(['Low', 'Moderate', 'High'], n, p=[0.3, 0.5, 0.2]),
    
    'bmi': np.round(np.random.uniform(18, 35, n), 1),
    
    'diabetes': np.random.choice(['Yes', 'No'], n, p=[0.18, 0.82]),
    
    'BP': np.random.choice(['Normal', 'Prehypertension', 'High'], n, p=[0.5, 0.3, 0.2]),
    
    'smoker': np.random.choice(['Yes', 'No'], n, p=[0.28, 0.72]),
    
    'alcohol': np.random.choice(['Low', 'Medium', 'High'], n, p=[0.4, 0.4, 0.2]),
    
    'region': np.random.choice(['North', 'East', 'West', 'South'], n, p=[0.3, 0.25, 0.25, 0.2]),
    
    # Conditions (realistic skew)
    'thyroid': np.random.choice(['Yes', 'No'], n, p=[0.12, 0.88]),
    'heart_disease': np.random.choice(['Yes', 'No'], n, p=[0.08, 0.92]),
    'asthma': np.random.choice(['Yes', 'No'], n, p=[0.22, 0.78]),
    'allergies': np.random.choice(['Yes', 'No'], n, p=[0.35, 0.65])
})

df.to_csv("insurance_extended.csv", index=False)
print("Dataset created with realistic imbalance!")
```
